# Notebook 01 — Data Extraction

## Objective

This notebook performs the data ingestion stage of the AirIntel pipeline.

The objectives are:

- Load the raw Indian Air Quality dataset.
- Inspect the dataset structure.
- Perform initial schema exploration.
- Generate a basic ingestion report.
- Save the raw dataset in Parquet format for downstream processing.


```python
## Import Libraries
import warnings
from pathlib import Path

import numpy as np
import pandas as pd

warnings.filterwarnings("ignore")

pd.set_option("display.max_columns", None)
pd.set_option("display.float_format", "{:.2f}".format)
```

```python
# Define project directories

PROJECT_ROOT = Path("..")

RAW_DATA = PROJECT_ROOT / "data" / "raw"
PROCESSED_DATA = PROJECT_ROOT / "data" / "processed"
REPORTS = PROJECT_ROOT / "reports"

PROCESSED_DATA.mkdir(parents=True, exist_ok=True)
REPORTS.mkdir(parents=True, exist_ok=True)
```

```python
# Load the raw dataset

df = pd.read_csv(
    RAW_DATA / "INDIA_AQI_COMPLETE_20251126.csv",
    low_memory=False
)

print(f"Rows    : {df.shape[0]:,}")
print(f"Columns : {df.shape[1]}")
```

```python
df.head()
```

```python
df.info()
```

```python
dataset_summary = pd.DataFrame({

    "Metric": [
        "Rows",
        "Columns",
        "Cities",
        "States"
    ],

    "Value": [
        len(df),
        df.shape[1],
        df["City"].nunique(),
        df["State"].nunique()
    ]
})

dataset_summary
```

```python
## Schema Inspection
schema = pd.DataFrame({

    "Column": df.columns,

    "Data Type": df.dtypes.astype(str),

    "Missing Values": df.isna().sum(),

    "Missing %": (
        df.isna().mean() * 100
    ).round(2)

})

schema
```

```python
## Memory Usage
memory_usage = df.memory_usage(deep=True).sum() / 1024**2

print(f"Dataset Memory Usage : {memory_usage:.2f} MB")
```

```python
## Date Range
print("Start Date :", df["Datetime"].min())
print("End Date   :", df["Datetime"].max())
```

```python
## Cities Covered
cities = pd.DataFrame({

    "City": sorted(df["City"].unique())

})

cities
```

```python
## Pollutant Availability
pollutants = [
    "PM2_5_ugm3",
    "PM10_ugm3",
    "NO2_ugm3",
    "SO2_ugm3",
    "CO_ugm3",
    "O3_ugm3"
]

availability = pd.DataFrame({

    "Pollutant": pollutants,

    "Available Values": [
        df[col].notna().sum()
        for col in pollutants
    ],

    "Missing Values": [
        df[col].isna().sum()
        for col in pollutants
    ]
})

availability
```

```python
# Save the ingested dataset in Parquet format

df.to_parquet(
    PROCESSED_DATA / "raw_airintel.parquet",
    index=False
)

print("Raw dataset saved successfully.")
```

```python
## Generate Ingestion Report
ingestion_report = pd.DataFrame({

    "Metric": [
        "Rows",
        "Columns",
        "Cities",
        "States",
        "Start Date",
        "End Date",
        "Memory Usage (MB)"
    ],

    "Value": [
        len(df),
        df.shape[1],
        df["City"].nunique(),
        df["State"].nunique(),
        df["Datetime"].min(),
        df["Datetime"].max(),
        round(memory_usage, 2)
    ]
})

ingestion_report
```

```python
ingestion_report.to_csv(
    REPORTS / "ingestion_report.csv",
    index=False
)
```
