# Notebook 02 — Data Validation

## Objective

This notebook validates the extracted AirIntel dataset before any cleaning is performed.

The objectives are:

- Validate the dataset schema.
- Analyze missing values.
- Detect duplicate records.
- Verify data types.
- Validate timestamps.
- Check city and state consistency.
- Validate pollutant ranges.
- Generate a comprehensive validation report.

No modifications are made to the dataset in this notebook.

```python
# Import required libraries

import warnings
from pathlib import Path

import numpy as np
import pandas as pd

warnings.filterwarnings("ignore")

pd.set_option("display.max_columns", None)
pd.set_option("display.float_format", "{:.2f}".format)
```

```python
# Define project directories

PROJECT_ROOT = Path("..")

PROCESSED_DATA = PROJECT_ROOT / "data" / "processed"
REPORTS = PROJECT_ROOT / "reports"
```

```python
# Load the processed dataset

df = pd.read_parquet(
    PROCESSED_DATA / "raw_airintel.parquet"
)

df.head()
```

```python
# Summarize the dataset

dataset_summary = pd.DataFrame({

    "Metric": [
        "Rows",
        "Columns",
        "Cities",
        "States",
        "Date Range"
    ],

    "Value": [
        len(df),
        df.shape[1],
        df["City"].nunique(),
        df["State"].nunique(),
        f"{df['Datetime'].min()} → {df['Datetime'].max()}"
    ]

})

dataset_summary
```

```python
# Review dataset schema

schema = pd.DataFrame({

    "Column": df.columns,

    "Data Type": df.dtypes.astype(str),

    "Missing Values": df.isna().sum(),

    "Missing %": (
        df.isna().mean()*100
    ).round(2)

})

schema
```

```python
# Calculate missing values

missing_summary = pd.DataFrame({

    "Missing Values": df.isna().sum(),

    "Missing %": (
        df.isna().mean()*100
    ).round(2)

}).sort_values(
    "Missing %",
    ascending=False
)

missing_summary
```

```python
# Identify duplicate records

duplicate_count = df.duplicated().sum()

duplicate_summary = pd.DataFrame({

    "Metric":[
        "Duplicate Rows"
    ],

    "Value":[
        duplicate_count
    ]

})

duplicate_summary
```

```python
# Summarize feature datatypes

datatype_summary = pd.DataFrame(

    df.dtypes,
    columns=["Data Type"]

)

datatype_summary
```

```python
# Validate datetime information

datetime_summary = pd.DataFrame({

    "Metric":[
        "Start Date",
        "End Date",
        "Unique Timestamps"
    ],

    "Value":[
        df["Datetime"].min(),
        df["Datetime"].max(),
        df["Datetime"].nunique()
    ]

})

datetime_summary
```

```python
# Summarize observations for each city

city_summary = (

    df.groupby("City")

    .size()

    .reset_index(name="Records")

    .sort_values(
        "Records",
        ascending=False
    )

)

city_summary
```

```python
# Summarize observations for each state

state_summary = (

    df.groupby("State")

    .size()

    .reset_index(name="Records")

    .sort_values(
        "Records",
        ascending=False
    )

)

state_summary
```

```python
# Validate geographic coordinates

coordinate_summary = pd.DataFrame({

    "Metric":[
        "Minimum Latitude",
        "Maximum Latitude",
        "Minimum Longitude",
        "Maximum Longitude"
    ],

    "Value":[
        df["Latitude"].min(),
        df["Latitude"].max(),
        df["Longitude"].min(),
        df["Longitude"].max()
    ]

})

coordinate_summary
```

```python
# Summarize pollutant ranges

pollutants = [

    "PM2_5_ugm3",
    "PM10_ugm3",
    "NO2_ugm3",
    "SO2_ugm3",
    "CO_ugm3",
    "O3_ugm3"

]

range_summary = pd.DataFrame({

    "Minimum":[
        df[col].min()
        for col in pollutants
    ],

    "Maximum":[
        df[col].max()
        for col in pollutants
    ]

},
index=pollutants)

range_summary
```

```python
# Calculate dataset memory usage

memory_usage = (

    df.memory_usage(
        deep=True
    ).sum()

    /1024**2

)

print(
    f"Memory Usage : {memory_usage:.2f} MB"
)
```

```python
# Identify negative pollutant values

pollutants = [
    "PM2_5_ugm3",
    "PM10_ugm3",
    "NO2_ugm3",
    "SO2_ugm3",
    "CO_ugm3",
    "O3_ugm3"
]

negative_summary = pd.DataFrame({

    "Negative Values": [
        (df[col] < 0).sum()
        for col in pollutants
    ]

},
index=pollutants)

negative_summary
```

```python
# Summarize pollutant statistics

range_validation = df[pollutants].describe().T

range_validation
```

```python
# Validate city-state mapping

city_state_validation = (

    df.groupby("City")["State"]

    .nunique()

    .reset_index(name="Unique States")

)

city_state_validation
```

```python
# Identify inconsistent city-state mappings

invalid_city_state = city_state_validation[
    city_state_validation["Unique States"] > 1
]

invalid_city_state
```

```python
# Validate city coordinates

coordinate_validation = (

    df.groupby("City")[["Latitude", "Longitude"]]

    .agg(["min", "max"])

)

coordinate_validation
```

```python
# Check duplicate timestamps for each city

duplicate_timestamp = (

    df.groupby(["City", "Datetime"])

    .size()

    .reset_index(name="Count")

)

duplicate_timestamp = duplicate_timestamp[
    duplicate_timestamp["Count"] > 1
]

duplicate_timestamp
```

```python
# Identify constant columns

constant_columns = [

    col

    for col in df.columns

    if df[col].nunique() == 1

]

constant_columns
```

```python
# Identify low variance numerical features

numeric_columns = df.select_dtypes(include=np.number).columns

variance_summary = pd.DataFrame({

    "Variance": df[numeric_columns].var()

}).sort_values(
    "Variance"
)

variance_summary.head(20)
```

```python
# Calculate unique values for every feature

unique_summary = pd.DataFrame({

    "Unique Values": df.nunique()

}).sort_values(
    "Unique Values"
)

unique_summary
```

```python
# Calculate overall data quality score

quality_checks = {

    "Duplicate Rows":
        duplicate_count == 0,

    "Missing Values":
        df.isna().sum().sum() == 0,

    "Negative Pollutants":
        negative_summary["Negative Values"].sum() == 0,

    "City-State Mapping":
        len(invalid_city_state) == 0,

    "Constant Columns":
        len(constant_columns) == 0

}

quality_score = (

    sum(quality_checks.values())

    / len(quality_checks)

) * 100

quality_report = pd.DataFrame({

    "Validation Check": quality_checks.keys(),

    "Passed": quality_checks.values()

})

quality_report
```

```python
# Display overall quality score

print(f"Overall Data Quality Score : {quality_score:.1f}%")
```

## Final Validation Summary

```python
# Summarize validation results

validation_summary = pd.DataFrame({

    "Metric": [

        "Rows",

        "Columns",

        "Cities",

        "States",

        "Duplicate Rows",

        "Columns with Missing Values",

        "Negative Pollutant Values",

        "Constant Columns",

        "Quality Score",

        "Memory Usage (MB)"

    ],

    "Value": [

        len(df),

        df.shape[1],

        df["City"].nunique(),

        df["State"].nunique(),

        duplicate_count,

        (df.isna().sum() > 0).sum(),

        negative_summary["Negative Values"].sum(),

        len(constant_columns),

        round(quality_score, 2),

        round(memory_usage,2)

    ]

})

validation_summary
```

```python
# Save validation report

validation_summary.to_csv(

    REPORTS / "validation_report.csv",

    index=False

)

print("Validation report saved successfully.")
```
